```python
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
import seaborn as sns

# Load the cleaned dataset
df = pd.read_csv("cleaned_drone_data.csv")

# Set Seaborn theme for clean aesthetics
sns.set_theme(style="whitegrid")
```

```python
plt.figure(figsize=(10, 5))
sns.lineplot(
    data=df.head(50),
    x=df.head(50).index,
    y="Altitude (meters)",
    color="darkblue",
    linewidth=2,
    marker="o",
)
plt.title("Flight Altitude Trajectory (First 50 Flights)")
plt.xlabel("Flight Sequence Index")
plt.ylabel("Altitude (meters)")
plt.show()


#no clear pattern
```

```python
plt.figure(figsize=(8, 5))
sns.histplot(df["Battery Remaining (%)"], bins=20, kde=True, color="seagreen")
plt.title("Distribution of Remaining Battery Levels (%)")
plt.xlabel("Battery Remaining (%)")
plt.ylabel("Frequency / Count")
plt.show()

#frequecy is higher at higher battery levels and falls at lower, though it ocillates up and down as well about every 10% or so
```

```python
plt.figure(figsize=(8, 5))
sns.boxplot(
    data=df, x="Drone Size", y="Battery Remaining (%)", palette="muted"
)
plt.title("Battery Retention Comparison Across Drone Sizes")
plt.xlabel("Drone Size")
plt.ylabel("Battery Remaining (%)")
plt.show()
```

```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

# Load dataset and clean column names
df = pd.read_csv("cleaned_drone_data.csv")
df.columns = df.columns.str.strip()

# Set theme
sns.set_theme(style="whitegrid")

# --- Fixed Scatter Plot: Flight Duration vs. Distance Flown ---
plt.figure(figsize=(9, 5))
sns.scatterplot(
    data=df,
    x="Flight Duration (minutes)",
    y="Distance Flown (km)",
    hue="Drone Size",
    palette="Set1",
    alpha=0.8,
    s=70,
)
plt.title("Flight Duration vs. Distance Flown by Drone Size")
plt.xlabel("Flight Duration (minutes)")
plt.ylabel("Distance Flown (km)")
plt.legend(bbox_to_anchor=(1.05, 1), loc="upper left")
plt.tight_layout()
plt.show()
#both are generally linearly dependent, howvever we can note that small drones are generally used for smaller flights, while medium are 
#used in almost all cases and have a varied, it would seem that drone size is directly related to the size od the task it is assigned.
```
